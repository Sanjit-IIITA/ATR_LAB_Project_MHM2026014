```python
from datasets import load_dataset
from rank_bm25 import BM25Okapi
import pandas as pd
import re

dataset_name = "carlfeynman/Bharat_NanoSCIDOCS_hi"

corpus = load_dataset(dataset_name, "corpus")["train"]
queries = load_dataset(dataset_name, "queries")["train"]
qrels = load_dataset(dataset_name, "qrels")["train"]

def tokenize(text):
    return re.findall(r"\w+", text.lower())

documents = [tokenize(row["text"]) for row in corpus]
bm25 = BM25Okapi(documents)

print("BM25 model created successfully")
print("Documents indexed:", len(documents))
print("Queries available:", len(queries))

sample_query = queries[0]["text"]
scores = bm25.get_scores(tokenize(sample_query))
top_indices = scores.argsort()[::-1][:5]

print("\nSample query:")
print(sample_query)

print("\nTop 5 retrieved documents:")
for rank, index in enumerate(top_indices, start=1):
    print(f"\nRank {rank} | Score: {scores[index]:.4f}")
    print(corpus[int(index)]["text"][:300])
```

```python
import numpy as np
import pandas as pd

corpus_ids = [row["_id"] for row in corpus]
query_ids = [row["_id"] for row in queries]

relevant_docs = {}

for row in qrels:
    relevant_docs.setdefault(row["query-id"], set()).add(row["corpus-id"])

results = []

for query in queries:
    query_id = query["_id"]
    query_text = query["text"]

    scores = bm25.get_scores(tokenize(query_text))
    ranked_indices = np.argsort(scores)[::-1]

    ranked_doc_ids = [corpus_ids[int(i)] for i in ranked_indices]
    relevant = relevant_docs.get(query_id, set())

    top5 = ranked_doc_ids[:5]
    top10 = ranked_doc_ids[:10]

    recall5 = len(set(top5) & relevant) / len(relevant)
    recall10 = len(set(top10) & relevant) / len(relevant)

    reciprocal_rank = 0

    for rank, doc_id in enumerate(top10, start=1):
        if doc_id in relevant:
            reciprocal_rank = 1 / rank
            break

    results.append({
        "query_id": query_id,
        "query": query_text,
        "relevant_count": len(relevant),
        "Recall@5": recall5,
        "Recall@10": recall10,
        "MRR@10": reciprocal_rank
    })

results_df = pd.DataFrame(results)

print("BM25 EVALUATION RESULTS")
print(results_df[["Recall@5", "Recall@10", "MRR@10"]].mean())

print("\nFIRST 10 QUERY-LEVEL RESULTS")
print(results_df.head(10).to_string(index=False))

results_df.to_csv("bm25_query_results.csv", index=False)

print("\nSaved: bm25_query_results.csv")
```

```python
import numpy as np
import torch
import pandas as pd
from sentence_transformers import SentenceTransformer

model_name = "intfloat/multilingual-e5-small"

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

dense_model = SentenceTransformer(model_name, device=device)

document_texts = [
    "passage: " + row["text"]
    for row in corpus
]

query_texts = [
    "query: " + row["text"]
    for row in queries
]

document_embeddings = dense_model.encode(
    document_texts,
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True,
    convert_to_numpy=True
)

query_embeddings = dense_model.encode(
    query_texts,
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True,
    convert_to_numpy=True
)

print("\nDense embeddings created successfully")
print("Document embeddings:", document_embeddings.shape)
print("Query embeddings:", query_embeddings.shape)
```

```python
dense_scores = query_embeddings @ document_embeddings.T

dense_results = []

for i, query in enumerate(queries):
    query_id = query["_id"]
    relevant = relevant_docs[query_id]

    ranked_indices = np.argsort(dense_scores[i])[::-1]
    ranked_doc_ids = [corpus_ids[int(j)] for j in ranked_indices]

    top5 = ranked_doc_ids[:5]
    top10 = ranked_doc_ids[:10]

    recall5 = len(set(top5) & relevant) / len(relevant)
    recall10 = len(set(top10) & relevant) / len(relevant)

    reciprocal_rank = 0

    for rank, doc_id in enumerate(top10, start=1):
        if doc_id in relevant:
            reciprocal_rank = 1 / rank
            break

    dense_results.append({
        "query_id": query_id,
        "query": query["text"],
        "relevant_count": len(relevant),
        "Recall@5": recall5,
        "Recall@10": recall10,
        "MRR@10": reciprocal_rank
    })

dense_results_df = pd.DataFrame(dense_results)

print("MULTILINGUAL DENSE RETRIEVAL RESULTS")
print(dense_results_df[["Recall@5", "Recall@10", "MRR@10"]].mean())

print("\nFIRST 10 QUERY-LEVEL RESULTS")
print(dense_results_df.head(10).to_string(index=False))

dense_results_df.to_csv("dense_query_results.csv", index=False)

print("\nSaved: dense_query_results.csv")
```

```python
comparison = results_df.merge(
    dense_results_df,
    on="query_id",
    suffixes=("_BM25", "_Dense")
)

metrics = ["Recall@5", "Recall@10", "MRR@10"]

print("AVERAGE PERFORMANCE")

for metric in metrics:
    bm25_avg = comparison[f"{metric}_BM25"].mean()
    dense_avg = comparison[f"{metric}_Dense"].mean()

    print(f"\n{metric}")
    print("BM25:", round(bm25_avg, 4))
    print("Dense:", round(dense_avg, 4))
    print("Difference:", round(dense_avg - bm25_avg, 4))

print("\nQUERY-LEVEL WINS, TIES AND LOSSES")

for metric in metrics:
    difference = (
        comparison[f"{metric}_Dense"] -
        comparison[f"{metric}_BM25"]
    )

    print(f"\n{metric}")
    print("Dense wins:", int((difference > 0).sum()))
    print("Ties:", int((difference == 0).sum()))
    print("BM25 wins:", int((difference < 0).sum()))

comparison["MRR_difference"] = (
    comparison["MRR@10_Dense"] -
    comparison["MRR@10_BM25"]
)

print("\nTOP 5 DENSE RETRIEVAL IMPROVEMENTS")

print(
    comparison.sort_values("MRR_difference", ascending=False)[
        ["query_BM25", "MRR@10_BM25", "MRR@10_Dense", "MRR_difference"]
    ].head(5).to_string(index=False)
)

print("\nTOP 5 DENSE RETRIEVAL REGRESSIONS")

print(
    comparison.sort_values("MRR_difference")[
        ["query_BM25", "MRR@10_BM25", "MRR@10_Dense", "MRR_difference"]
    ].head(5).to_string(index=False)
)

comparison.to_csv("retrieval_comparison.csv", index=False)

print("\nSaved: retrieval_comparison.csv")
```

```python
query_index = {
    row["_id"]: i for i, row in enumerate(queries)
}

def show_results(query_id, query_text):
    i = query_index[query_id]
    relevant = relevant_docs[query_id]

    bm25_query_scores = bm25.get_scores(tokenize(query_text))
    bm25_order = np.argsort(bm25_query_scores)[::-1][:3]

    dense_order = np.argsort(dense_scores[i])[::-1][:3]

    print("\n" + "=" * 80)
    print("QUERY:", query_text)
    print("Relevant documents:", len(relevant))

    print("\nBM25 TOP 3")

    for rank, idx in enumerate(bm25_order, start=1):
        doc_id = corpus_ids[int(idx)]
        status = "RELEVANT" if doc_id in relevant else "NOT RELEVANT"

        print(f"\nRank {rank} | {status}")
        print(corpus[int(idx)]["text"][:350])

    print("\nDENSE RETRIEVAL TOP 3")

    for rank, idx in enumerate(dense_order, start=1):
        doc_id = corpus_ids[int(idx)]
        status = "RELEVANT" if doc_id in relevant else "NOT RELEVANT"

        print(f"\nRank {rank} | {status}")
        print(corpus[int(idx)]["text"][:350])


improvements = comparison.sort_values(
    "MRR_difference", ascending=False
).head(3)

regressions = comparison.sort_values(
    "MRR_difference"
).head(3)

print("DENSE RETRIEVAL IMPROVEMENTS")

for _, row in improvements.iterrows():
    show_results(row["query_id"], row["query_BM25"])

print("\n\nBM25 IMPROVEMENTS")

for _, row in regressions.iterrows():
    show_results(row["query_id"], row["query_BM25"])
```

```python
def get_top_result(method, query_id):
    i = query_index[query_id]
    relevant = relevant_docs[query_id]

    if method == "BM25":
        scores = bm25.get_scores(tokenize(queries[i]["text"]))
        ranked = np.argsort(scores)[::-1]
    else:
        ranked = np.argsort(dense_scores[i])[::-1]

    idx = int(ranked[0])
    doc_id = corpus_ids[idx]

    return {
        "top_document_id": doc_id,
        "top_document_text": corpus[idx]["text"][:250],
        "top1_relevant": doc_id in relevant
    }


error_rows = []

for _, row in comparison.iterrows():
    bm25_top = get_top_result("BM25", row["query_id"])
    dense_top = get_top_result("Dense", row["query_id"])

    if bm25_top["top1_relevant"] and not dense_top["top1_relevant"]:
        pattern = "BM25 only"
    elif dense_top["top1_relevant"] and not bm25_top["top1_relevant"]:
        pattern = "Dense only"
    elif dense_top["top1_relevant"] and bm25_top["top1_relevant"]:
        pattern = "Both"
    else:
        pattern = "Neither"

    error_rows.append({
        "query": row["query_BM25"],
        "BM25_top1_relevant": bm25_top["top1_relevant"],
        "Dense_top1_relevant": dense_top["top1_relevant"],
        "top1_pattern": pattern,
        "BM25_document": bm25_top["top_document_text"],
        "Dense_document": dense_top["top_document_text"]
    })

error_analysis_df = pd.DataFrame(error_rows)

print("TOP-1 RELEVANCE PATTERNS")
print(error_analysis_df["top1_pattern"].value_counts())

print("\nQUERY-LEVEL ERROR ANALYSIS")
print(
    error_analysis_df[
        ["query", "BM25_top1_relevant", "Dense_top1_relevant", "top1_pattern"]
    ].to_string(index=False)
)

error_analysis_df.to_csv("error_analysis.csv", index=False)

print("\nSaved: error_analysis.csv")
```
